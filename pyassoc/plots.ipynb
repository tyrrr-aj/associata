```python
n_cores = list(range(1, 17))
markers = ['o', 's', '^', 'd']
cases = ['Balanced', 'Big VNG', 'Little VNG', 'Categorical']
titles = {
    'Balanced': '(1) Numerical VNGs of similar size',
    'Big VNG': '(2) Numerical VNGs, one of them with large number of VNs',
    'Little VNG': '(3) Numerical VNGs, one of them with small number of VNs',
    'Categorical': '(4) Mixed VNGs: half of them numeriacl, half categorical'
}

fig, axs = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Average speedup achieved for varying AGDNN setup', fontsize=18)
fig.subplots_adjust(top=0.98)

for idx, case in enumerate(cases):
    row = idx // 2
    col = idx % 2
    ax = axs[row, col]
    
    for exp_size, marker in zip(results[case], markers):
        single_core_time = exp_size['inference_times'][0]
        speedup = [100 * single_core_time / time for time in exp_size['inference_times']]
        ax.plot(n_cores, speedup, label=f'{exp_size["structure_size"]} nodes', marker=marker)
    
    ax.set_xlabel('Number of CPU cores', fontsize=14)
    ax.set_ylabel('Speed-up [%]', fontsize=14)
    ax.set_title(titles[case], fontsize=16)
    ax.legend(fontsize=12)
    ax.grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.98])
plt.savefig('results-strong-speedup.eps', format='eps', dpi=1000)
plt.show()
```
